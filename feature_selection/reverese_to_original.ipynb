```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('/content/gurgaon_properties_missing_value_imputation.csv')
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.head()
```

```python
df.shape
```

```python
train_df = df.drop(columns=['society','price_per_sqft'])
```

```python
train_df.head()
```

```python
def categorize_luxury(score):
    if 0 <= score < 50:
        return "Low"
    elif 50 <= score < 150:
        return "Medium"
    elif 150 <= score <= 175:
        return "High"
    else:
        return None  # or "Undefined" or any other label for scores outside the defined bins
```

```python
train_df['luxury_category'] = train_df['luxury_score'].apply(categorize_luxury)
```

```python
def categorize_floor(floor):
    if 0 <= floor <= 2:
        return "Low Floor"
    elif 3 <= floor <= 10:
        return "Mid Floor"
    elif 11 <= floor <= 51:
        return "High Floor"
    else:
        return None  # or "Undefined" or any other label for floors outside the defined bins
```

```python
train_df['floor_category'] = train_df['floorNum'].apply(categorize_floor)
```

```python
train_df.drop(columns=['floorNum','luxury_score'],inplace=True)
```

```python
export_df = train_df.drop(columns=['pooja room', 'store room', 'others'])
```

```python
export_df.to_csv('gurgaon_properties_post_feature_selection_2.0.csv', index=False)
```
