```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('/content/gurgaon_properties_missing_value_imputation.csv')
```

```python
df.shape
```

```python
df.head()
```

```python
train_df = df.drop(columns=['society','price_per_sqft'])
```

```python
numeric_train_df = train_df.select_dtypes(include='number')
plt.figure(figsize=(14, 10))
sns.heatmap(numeric_train_df.corr(),cmap='coolwarm',annot=True,fmt='.2f',center=0,linewidths=0.5)
plt.title('Correlation Heatmap', fontsize=16, fontweight='bold')
plt.show()
```
